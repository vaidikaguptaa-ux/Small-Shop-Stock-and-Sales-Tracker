```python
products={"chocolate":{"price":20,"stock":50},"biscuits":{"price":10,"stock":30},"juice":{"price":40,"stock":15}}
print(products)
print(products["chocolate"]["price"])
print(products["juice"]["stock"])
```

```python
import json
import os

DATA_FILE = "shop_data.json"

# Load saved data if available
if os.path.exists(DATA_FILE):
    with open(DATA_FILE, "r") as file:
        saved_data = json.load(file)

    products = saved_data["products"]
    total_sales = saved_data["total_sales"]
else:
    total_sales = 0

op = 0

while op != 5:
    print("\nSMALL SHOP STOCK AND SALES TRACKER")
    print("----------------------------------")
    print("1. View stock")
    print("2. Record a sale")
    print("3. Show total sales")
    print("4. Save data")
    print("5. Exit")

    op = int(input("Choose an option: "))


    if op == 1:
        print("\nCURRENT STOCK")

        for name in products:
            print("Product:", name)
            print("Price: Rs.", products[name]["price"])
            print("Stock:", products[name]["stock"])
            print("--------------------")

    elif op == 2:
        product_name = input(
            "Enter product name: "
        ).strip().lower()

        if product_name in products:
            quantity = int(input("Enter quantity: "))

            if quantity > 0 and quantity <= products[product_name]["stock"]:
                price = products[product_name]["price"]
                bill = price * quantity

                products[product_name]["stock"] -= quantity
                total_sales += bill

                print("\nSale successful!")
                print("Product:", product_name)
                print("Quantity:", quantity)
                print("Total bill: Rs.", bill)
                print("Remaining stock:", products[product_name]["stock"])

            else:
                print("Invalid quantity or insufficient stock.")

        else:
            print("Product not found!")

    elif op == 3:
        print("Total sales: Rs.", total_sales)

    elif op == 4:
        saved_data = {
            "products": products,
            "total_sales": total_sales
        }

        with open(DATA_FILE, "w") as file:
            json.dump(saved_data, file, indent=4)

        print("Data saved successfully!")

    elif op == 5:
        print("Thank you for using our shop tracker!")

    else:
        print("Invalid option. Please choose 1 to 5.")
```
